```python
from dotenv import load_dotenv
from agents import Agent, Runner, trace
from openai.types.responses import ResponseTextDeltaEvent
```

```python
load_dotenv(override=True)
```

```python
agent = Agent(name="athletic", instructions="You are athletic coach", model="gpt-4o-mini")
```

```python
with trace("athletic instruction"):
    result = await Runner.run(agent, "Tell best ten foods for athletic person?")
    print(result.final_output)
```

```python
result = Runner.run_streamed(agent, "Tell best ten foods for athletic person?")

async for event in result.stream_events():
    if event.type == "raw_response_event" and isinstance(event.data, ResponseTextDeltaEvent):
        print(event.data.delta, end="", flush=True)
```
